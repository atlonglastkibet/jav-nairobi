```python
import geopandas as gpd
import pandas as pd
import folium
from folium import Choropleth, plugins
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import numpy as np

# Load GeoDataFrame (use .gpkg for geometry; fallback to CSV + WKT parse if needed)
gdf = gpd.read_file('/home/dataopske/Desktop/jav/data/processed/ward_temporal_with_mobility.gpkg')
# Or for CSV: gdf = gpd.GeoDataFrame(pd.read_csv('...csv'), geometry=gpd.GeoSeries.from_wkt(df['geometry']))

# Filter to target hours
target_hours = [6, 9, 15]
gdf_filtered = gdf[gdf['hour'].isin(target_hours)].copy()

print(f"Shape: {gdf_filtered.shape} | Unique wards: {gdf_filtered['ward'].nunique()}")
gdf_filtered.explore()  # Quick basemap preview
```

```python
def plot_choropleth(gdf, column, hour, cmap='YlOrRd', title=''):
    m = folium.Map(location=[-1.2921, 36.8219], zoom_start=11, tiles='OpenStreetMap')  # Nairobi center
    
    # Choropleth
    Choropleth(
        gdf[gdf['hour'] == hour].set_index('ward'),
        data=gdf[gdf['hour'] == hour][column],
        columns=[gdf[gdf['hour'] == hour].index, gdf[gdf['hour'] == hour][column]],
        key_on='feature.properties.ward',  # Assumes 'ward' is in GeoJSON props
        fill_color=cmap,
        fill_opacity=0.7,
        line_opacity=0.2,
        legend_name=column,
        nan_fill_color='white'
    ).add_to(m)
    
    # Popups
    for idx, row in gdf[gdf['hour'] == hour].iterrows():
        folium.Marker(
            row.geometry.centroid.coords[0],
            popup=f"Ward: {row['ward']}<br>{column}: {row[column]:.1f}",
            icon=folium.Icon(color='blue', icon='info-sign')
        ).add_to(m)
    
    m.save(f'{title}_{hour}.html')
    return m

# Example: Trips at 6,9,15
for h in target_hours:
    plot_choropleth(gdf_filtered, 'outgoing_trips', h, title='Outgoing Trips')
    plot_choropleth(gdf_filtered, 'avg_speed', h, cmap='RdYlGn_r', title='Avg Speed')  # Reverse for speed (green=fast)
```
